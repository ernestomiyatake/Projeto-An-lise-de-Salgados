```python
import pandas as pd
pd.plotting.register_matplotlib_converters()
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
print("Setup Complete")
```

```python
# Criando uma tabela de salgados e seus respectivos pedidos
dados = {
    'Salgados': ['Coxinha', 'Risóles', 'Bolinha de Queijo', 'Enrolado de Salsicha', 'Kibe', 'Espeto Salada'], 
    'Pedidos': [1290, 790, 950, 670, 880, 1300]
}

# Criando um Data Frame:
df = pd.DataFrame(dados)

# Configurando o gráfico de colunas
plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Salgados', y='Pedidos', hue='Salgados', palette='Oranges_r', legend=False)

plt.title('Total de Pedidos por Tipo de Salgado', fontsize=14, fontweight='bold')
plt.xlabel('Salgados', fontsize=12)
plt.ylabel('Pedidos', fontsize=12)

# Rotacionando os nomes do eixo X para não embolarem
plt.xticks(rotation=15)

plt.show()
```
